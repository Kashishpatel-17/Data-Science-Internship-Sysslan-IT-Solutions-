```python
import pandas as pd
df = pd.read_csv("Dataset1.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df["Arrival_time"] = pd.to_datetime(
    df["Arrival_time"],
    format="%H:%M:%S",
    errors="coerce"
)

df["Departure_Time"] = pd.to_datetime(
    df["Departure_Time"],
    format="%H:%M:%S",
    errors="coerce"
)
```

```python
df[["Arrival_time", "Departure_Time"]].head()
```

```python
journey_duration = df.groupby("Train_No").agg(
    Start_Time=("Arrival_time", "min"),
    End_Time=("Departure_Time", "max")
)

journey_duration["Journey_Duration_Hours"] = (
    journey_duration["End_Time"] - journey_duration["Start_Time"]
).dt.total_seconds() / 3600

journey_duration.head()
```

```python
total_distance = df.groupby("Train_No")["Distance"].max()
```

```python
num_stops = df.groupby("Train_No")["Station_Code"].count()
```

```python
train_features = pd.concat(
    [total_distance, num_stops, journey_duration["Journey_Duration_Hours"]],
    axis=1
)

train_features.columns = [
    "Total_Distance",
    "Number_of_Stops",
    "Journey_Duration_Hours"
]

train_features.head()
```
